# Problem Statement

## Business Context

**AuroraStay Hotels** is a mid-sized hotel group operating **18 business and leisure properties** across major cities in the USA, adding **3 to 4 new properties every year**.

Hotel staff handle around **1,200 guest queries per week** about:

- Room types
- Check-in/check-out
- Cancellation policies
- Refunds
- Breakfast
- Parking
- Pets
- Amenities
- Additional charges

Simple queries can be answered using existing FAQs. However, around **40% of queries are complex** and require information from multiple documents, such as:

- Booking terms
- Cancellation policies
- Room information
- Guest guidelines

Manually combining this information takes **8 to 12 minutes per query** and results in a **15% to 20% error rate** because answers may be incomplete or incorrect.

As the hotel group grows, the number of properties and property-specific policies also increases. With **45% annual staff turnover**, institutional knowledge is constantly lost.

## Key Risks

- **Faithfulness:** Avoid unsupported information.
- **Retrieval Completeness:** Retrieve all relevant information required to answer the query.
- **Retrieval Precision:** Retrieve relevant source chunks while minimizing irrelevant information.
- **Answer Relevancy:** Directly answer the guest's question.

A **RAG system** can retrieve relevant information from these documents and help staff provide **accurate, source-grounded answers**.

## Objective

The objective is to build a **POC for a user interface** for AuroraStay Hotels where hotel staff can enter natural-language questions and receive answers based on the hotel's documents.

The application should:

- Provide a simple UI where staff can enter their questions.
- Retrieve the most relevant information from the hotel's document collection.
- Generate an answer grounded **only in the retrieved information**.
- Display the supporting source information along with the answer.
- **Reduce the number of queries that require manual lookup or escalation**, rather than focusing on reducing the time spent handling each query.

Since this is a **POC**, the application will initially be **internal-facing**, allowing the hotel team to test it extensively, ensure that user queries are answered consistently, and validate the **accuracy and reliability** of the system before deploying it for broader use.

> **Note:** The objective at this stage is **not to reduce query-handling time**. The primary objective is to determine whether the system can reliably answer a meaningful proportion of queries without requiring manual effort.

## Success Criteria

| **Business Objective** | **Technical Metric** | **Threshold** |
|---|---|---:|
| Avoid unsupported information | Faithfulness | >= 0.85 |
| Retrieve all relevant information | Contextual Recall | >= 0.80 |
| Retrieve relevant chunks | Contextual Precision | >= 0.75 |
| Answer the question asked | Answer Relevancy | >= 0.75 |
| Reduce the number of queries requiring manual searches for policy information | Queries Confidently Answered | >= 50% |

**Composite Score = 0.35 × Faithfulness + 0.25 × Contextual Recall + 0.20 × Contextual Precision + 0.20 × Answer Relevancy**

The weights for the composite score are **informed by past evaluation results and the relative importance of each metric to the business objectives**. Faithfulness is given the highest weight because avoiding unsupported or misleading information is the most critical requirement in this use case.

The pipeline routes every generated answer through a **scoring stage**. Based on the composite score, the answer is routed to one of two actions:

1. **Auto Answer (score >= 75%)**  
   The generated answer and its supporting source chunks are considered reliable enough to present directly to the representative for immediate use with the customer.

2. **Manual Lookup (score < 75%)**  
   The system output is not considered reliable enough to act on. The representative falls back to the existing process by looking up the answer manually or escalating it to the back-office helpdesk.

## Chunking

### Choosing the embedding model

Here, the embedding model chosen is **"HarishMaths/Hotel-Policy-Embedding"**, which is fine-tuned on hotel policy documents. Hence, it is expected to provide better retrieval quality compared to other base models.

````python
# define the model name
EMBED_MODEL = "HarishMaths/Hotel-Policy-Embedding"

# Load the domain-specific embedding model tokenizer and check its maximum token limit
tokenizer = AutoTokenizer.from_pretrained(EMBED_MODEL)

print(f"Model max tokens: {tokenizer.model_max_length}")
````

- The model's maximum token length tells us the upper bound on how much text can be encoded into a single embedding vector.
-  This limit directly affects how we configure the chunking strategy in the next step.
- Chunks that exceed this limit would be silently truncated, potentially losing important information.

Before chunking, it helps to see how the tokenizer actually breaks text into tokens. This gives us a sense of the token-to-word ratio for hotel policy language.

````python
# Test tokenization with a property-specific hotel policy question.
sample = "What are the pet fees at ASH-CHI Loop?"

tokens = tokenizer.tokenize(sample)
print(f"Sample: '{sample}'")
print(f"Tokens ({len(tokens)}): {tokens}")
````

- The sample sentence produces a specific number of tokens, and we can see how individual words get split into subwords.
- Knowing this ratio helps us reason about chunk sizes.
  - For example, a 512-token chunk will typically hold roughly 350 to 400 words of hotel policy text, which is enough to capture a complete policy section in most cases.

### Define the chunker

**Guidelines**

- Define a text splitter of your choice.
- The text splitter **must be assigned to the variable `splitter`**, as this variable is used in the code that follows.

````python
# Count content tokens using the embedding model's own tokenizer.
# Disable truncation so the measured length is the actual length.
def token_length(text):
    return len(tokenizer.encode(
        text, add_special_tokens=False, truncation=False, verbose=False
    ))

# Reserve space for special tokens such as BERT's [CLS] and [SEP].
MODEL_MAX_TOKENS = int(tokenizer.model_max_length)
SPECIAL_TOKENS = tokenizer.num_special_tokens_to_add(pair=False)
CHUNK_SIZE = min(120, MODEL_MAX_TOKENS - SPECIAL_TOKENS)
CHUNK_OVERLAP = min(24, CHUNK_SIZE // 5)
assert 0 < CHUNK_OVERLAP < CHUNK_SIZE

# Preserve section headings, labeled table rows, and complete parent context.
from policy_retrieval import PolicyTextSplitter
splitter = PolicyTextSplitter(
    tokenizer=tokenizer, chunk_size=CHUNK_SIZE, chunk_overlap=CHUNK_OVERLAP,
)

print(f"Model input limit: {MODEL_MAX_TOKENS} tokens")
print(f"Special tokens per chunk: {SPECIAL_TOKENS}")
print(f"Chunk size: {CHUNK_SIZE} content tokens; overlap target: {CHUNK_OVERLAP} tokens")
````

### Chunking the documents

````python
# Export the documents; the splitter labels table cells and preserves parent context.
raw_docs = [
    LCDocument(
        page_content=doc.export_to_markdown(compact_tables=True),
        metadata={"source": path},
    )
    for path, doc in docs.items()
]

lc_docs = splitter.split_documents(raw_docs)
print(f"Total chunks: {len(lc_docs)}")

# Validate actual model inputs, including special tokens, without truncation.
chunk_token_counts = [token_length(chunk.page_content) for chunk in lc_docs]
model_input_token_counts = [
    len(tokenizer.encode(
        chunk.page_content, add_special_tokens=True, truncation=False, verbose=False
    ))
    for chunk in lc_docs
]
assert lc_docs and all(chunk.page_content.strip() for chunk in lc_docs)
assert max(chunk_token_counts) <= CHUNK_SIZE
assert max(model_input_token_counts) <= MODEL_MAX_TOKENS
assert {source for chunk in lc_docs for source in chunk.metadata["sources"]} == set(docs)

chunk_stats = pd.DataFrame({
    "source": [chunk.metadata["source"] for chunk in lc_docs],
    "content_tokens": chunk_token_counts,
    "model_input_tokens": model_input_token_counts,
})
display(chunk_stats.groupby("source", sort=False).agg(
    chunks=("content_tokens", "size"),
    min_content_tokens=("content_tokens", "min"),
    max_content_tokens=("content_tokens", "max"),
    max_model_input_tokens=("model_input_tokens", "max"),
))
print("All chunks fit the tokenizer limit, including special tokens; all source PDFs are represented.")

print("Structured chunk audit:", splitter.audit)
````

Each PDF is split into complete paragraphs and individually labeled table rows. Section headings remain attached to each embedding input. Identical policy content is indexed once, retaining all source filenames.

Embedding inputs remain within the model limit. The full parent paragraph or row is retained as metadata and restored for answer generation, preserving conditions and exceptions.

### Retrieval System

The retrieval function combines keyword matching with FAISS similarity ranks, respects explicit property codes and rate plans, and removes duplicate parent passages. It returns up to `k` relevant contexts with source metadata; it can return fewer when additional passages do not match the subject.


````python
# Combine keyword matching and FAISS, returning up to k distinct policy contexts.
from policy_retrieval import retrieve_policy_documents

def retrieve(query: str, k: int = 4, vectorstore=None) -> list[str]:
    results = retrieve_policy_documents(vectorstore, query, k=k)
    return [
        f"Content:\n{doc.page_content}\n\nMetadata:\n" + str({
            "source": doc.metadata["source"],
            "sources": doc.metadata.get("sources", [doc.metadata["source"]]),
            "section": doc.metadata.get("heading", ""),
        })
        for doc in results
    ]
````

The `retrieve` function returns distinct parent contexts with source metadata. Keyword and semantic rankings are combined using reciprocal rank fusion. Exact property matches are prioritized, and topic/rate-plan filters limit unrelated results. `k` is a maximum context count, defaulting to 4.


Let us test the retrieval system with a sample query about pet charges at a specific property.

````python
# Verify that exact-property facts are prioritized by hybrid retrieval.
query = "What are the pet fees at ASH-CHI Loop?"
RETRIEVAL_TEST_K = 4
retrieved_chunks = retrieve(query, k=RETRIEVAL_TEST_K, vectorstore=vectorstore)
for i, chunk in enumerate(retrieved_chunks, start=1):
    print(f"\n--- Result {i} ---")
    print(chunk)
expected_facts = ("ASH-CHI Loop", "$50 per night", "$150 stay cap")
assert all(fact in retrieved_chunks[0] for fact in expected_facts)
print("The first context contains the requested property's pet fee and stay cap.")
print(f"Returned {len(retrieved_chunks)} distinct parent contexts (limit {RETRIEVAL_TEST_K}).")
````

- The retrieved chunks contain information about pet policies and charges.

- If the chunks are relevant to the query, it means the embedding model and vector store are working together correctly.

### Defining the generation system

**Guidelines**

- Define your own `rules` to guide how the LLM should answer hotel policy questions.
- Your rules should explain **how the retrieved information should be used** and **how the final answer should be written**.
- Consider instructions such as using only the retrieved information, avoiding assumptions, including relevant conditions or exceptions, and clearly handling cases where the information is insufficient.
- You can write multiple rules based on what you want the RAG application to do.
- Experiment with different rules and observe how they affect the generated answers.

````python
# Define the generation function that prompts the LLM with the retrieved context.
def generate(query, retrieved_chunks, model):
    rules = """
    - Use only the retrieved policy information to answer the question.
    - Match the requested property exactly. Preserve fee amounts, currencies,
      per-night/per-stay units, caps, conditions, and exceptions.
    - If the information needed to answer is present, give a direct, concise answer.
      Other properties' rows do not override the requested property's row.
    - Cite the source filename from the metadata of the chunk containing each
      fact used in the answer. Do not cite a different chunk about another property.
    - If the retrieved information is insufficient or conflicting, explain what
      is missing or conflicting and ask for verification with the current property
      profile or property team. Do not invent policy details or use outside knowledge.
    """
    prompt = f"""
    You are an AI assistant answering questions about hotel policies.

    User Query:
    {query}

    Retrieved Policy Information:
    {retrieved_chunks}

    Rules:
    {rules}

    Answer:
    """
    return model.invoke(prompt)
````

### Defining the RAG Function

- The `rag` function ties the retrieval and generation steps together into a single callable pipeline.
- It creates a fresh LLM instance with the specified parameters, retrieves the top-k chunks, and passes them to the generation function.
- It returns both the generated answer and the retrieved chunks, which is useful for evaluation and for displaying source information in the UI.

````python
# Combine retrieval and generation, returning answer text and source excerpts.
def rag(
    query: str,
    k: int = 4,
    model_name: str = "gpt-4o-mini",
    temperature: float = 0.0,
    top_p: float = 1.0,
    max_tokens: int = 512,
    vectorstore=None,
):
    if vectorstore is None:
        raise ValueError("Provide the initialized FAISS vectorstore.")

    model = ChatOpenAI(
        model=model_name,
        temperature=temperature,
        top_p=top_p,
        max_tokens=max_tokens,
        api_key=config["OPENAI_API_KEY"],
        base_url=config["OPENAI_API_BASE"].strip(),
        timeout=60,
        max_retries=0,
    )
    retrieved_chunks = retrieve(query=query, k=k, vectorstore=vectorstore)

    answer = generate(query=query, retrieved_chunks=retrieved_chunks, model=model)
    return answer.content, retrieved_chunks
````

## Defining Evaluation Metrics

Four metrics are defined, each corresponding to one of the risks identified in the problem statement.
- **Faithfulness** guards against misrepresentation
- **Contextual Recall** checks retrieval completeness
- **Contextual Precision** validates that retrieved chunks are relevant and verifiable
- **Answer Relevancy** ensures the response addresses the actual question.

````python
# Faithfulness
# Checks whether the generated answer is factually supported by
# the information retrieved from the knowledge base.
faithfulness_metric = GEval(
    name="Faithfulness",

    # Defines what the evaluator should judge.
    criteria=(
        "Determine whether the actual output is factually supported by "
        "the provided retrieval context."
    ),

    # Step-by-step instructions given to the evaluator.
    evaluation_steps=[
        # First, identify the individual factual claims made in the answer.
        "Break the actual output into individual factual claims.",

        # Check each claim against the retrieved context.
        "Check whether each claim is supported by the retrieval context.",

        # Lower the score if the answer contains unsupported or conflicting claims.
        "Penalize claims that are unsupported or contradicted by the retrieval context.",

        # Differences in wording should not affect the score if the meaning is supported.
        "Do not penalize differences in wording when the meaning is supported by the context.",

        # Give a high score when the important claims are supported.
        "Give a higher score when all important claims are supported by the context."
    ],

    # Specifies which parts of the test case are used for evaluation.
    evaluation_params=[
        SingleTurnParams.ACTUAL_OUTPUT,
        SingleTurnParams.RETRIEVAL_CONTEXT
    ],

    # Minimum score required for the test case to pass.
    threshold=0.85,

    # LLM used to perform the evaluation.
    model=eval_model
)


# Contextual Recall
# Checks whether the retrieved context contains the information
# needed to produce the expected answer.
recall_metric = GEval(
    name="Contextual Recall",

    # Defines what the evaluator should judge.
    criteria=(
        "Determine whether the retrieval context contains the information "
        "needed to support the expected output."
    ),

    # Step-by-step instructions given to the evaluator.
    evaluation_steps=[
        # Identify the important claims that the expected answer should contain.
        "Break the expected output into its important factual claims.",

        # Check whether those claims can be found or supported by the retrieved context.
        "Check whether each important claim is present or supported by the retrieval context.",

        # Lower the score when important information is missing from retrieval.
        "Penalize the retrieval context when important information required for the expected output is missing.",

        # Wording differences should not affect the evaluation.
        "Do not penalize differences in wording or presentation.",

        # Give a high score when retrieval contains most or all required information.
        "Give a higher score when the retrieval context contains most or all information needed to produce the expected output."
    ],

    # Recall requires the expected answer and retrieved context.
    evaluation_params=[
        SingleTurnParams.EXPECTED_OUTPUT,
        SingleTurnParams.RETRIEVAL_CONTEXT
    ],

    # Minimum score required for the test case to pass.
    threshold=0.80,

    # LLM used to perform the evaluation.
    model=eval_model
)


# Contextual Precision
# Checks whether the retrieved context is relevant to the user's question
# and whether relevant information is prioritized over irrelevant information.
precision_metric = GEval(
    name="Contextual Precision",

    # Defines what the evaluator should judge.
    criteria=(
        "Determine whether the retrieved context is relevant to answering "
        "the input question, with relevant information prioritized over irrelevant information."
    ),

    # Step-by-step instructions given to the evaluator.
    evaluation_steps=[
        # Identify which parts of the retrieved context are useful for the question.
        "Identify the information in the retrieval context that is relevant to the input question.",

        # Check whether unnecessary or unrelated information was retrieved.
        "Check whether the retrieval context contains unnecessary or unrelated information.",

        # Lower the score when a large portion of the retrieved context is irrelevant.
        "Penalize retrievals that contain a large amount of irrelevant information.",

        # Give more credit when useful information appears prominently.
        "Give more credit when relevant information is prioritized.",

        # Give a high score when most retrieved information is useful.
        "Give a higher score when most of the retrieved context is useful for answering the question."
    ],

    # Precision evaluates the question against the retrieved context.
    evaluation_params=[
        SingleTurnParams.INPUT,
        SingleTurnParams.RETRIEVAL_CONTEXT
    ],

    # Minimum score required for the test case to pass.
    threshold=0.75,

    # LLM used to perform the evaluation.
    model=eval_model
)


# Answer Relevancy
# Checks whether the generated answer directly addresses the user's question
# without including unnecessary or unrelated information.
answer_relevancy_metric = GEval(
    name="Answer Relevancy",

    # Defines what the evaluator should judge.
    criteria=(
        "Determine whether the actual output directly addresses the input "
        "question and provides an appropriate response without unnecessary information."
    ),

    # Step-by-step instructions given to the evaluator.
    evaluation_steps=[
        # Check whether the answer directly responds to the question.
        "Check whether the actual output directly answers the input question.",

        # Check whether the information provided is relevant to the question.
        "Check whether the response contains information relevant to the question.",

        # Lower the score for unrelated or unnecessary information.
        "Penalize unnecessary, unrelated, or off-topic information.",

        # A short answer should not be penalized if it answers the question sufficiently.
        "Do not penalize concise answers when they adequately address the question.",

        # Give a high score when the answer is focused and directly relevant.
        "Give a higher score when the response is focused and directly relevant to the question."
    ],

    # Answer relevancy evaluates the question against the generated answer.
    evaluation_params=[
        SingleTurnParams.INPUT,
        SingleTurnParams.ACTUAL_OUTPUT
    ],

    # Minimum score required for the test case to pass.
    threshold=0.75,

    # LLM used to perform the evaluation.
    model=eval_model
)


# Store all evaluation metrics in a single list.
# This list can then be passed to DeepEval's evaluation function.
metrics = [
    faithfulness_metric,
    recall_metric,
    precision_metric,
    answer_relevancy_metric
]
````

Four evaluation metrics are defined, each targeting a specific aspect of RAG quality:

- **Faithfulness** (threshold >= 0.85): Checks whether the generated answer is factually supported by the retrieved context. Uses `ACTUAL_OUTPUT` and `RETRIEVAL_CONTEXT`.
- **Contextual Recall** (threshold >= 0.80): Checks whether the retrieved context contains the information needed to produce the expected answer. Uses `EXPECTED_OUTPUT` and `RETRIEVAL_CONTEXT`.
- **Contextual Precision** (threshold >= 0.75): Checks whether the retrieved chunks are relevant to the question and whether useful information is prioritized. Uses `INPUT` and `RETRIEVAL_CONTEXT`.
- **Answer Relevancy** (threshold >= 0.75): Checks whether the answer directly addresses the user's question without unnecessary information. Uses `INPUT` and `ACTUAL_OUTPUT`.

All four metrics use the GEval framework, which provides step-by-step evaluation instructions to the judge model. The thresholds match the success criteria defined in the problem statement.

## Building Evaluation Test Cases

The function below takes the evaluation DataFrame and runs the RAG pipeline on each question. It then packages the inputs, outputs, and ground-truth data into DeepEval test cases for scoring.

````python
def test_case_generator(df, vectorstore, k=4, temperature=0, top_p=1, max_tokens=512):
    required_columns = ["question", "expected_answer", "relevant_context"]
    if not set(required_columns).issubset(df.columns):
        raise ValueError("Evaluation data is missing required columns.")
    if df[required_columns].isna().any().any():
        raise ValueError("Evaluation data contains missing values.")
    if not all(df[column].map(lambda value: isinstance(value, str) and bool(value.strip())).all()
               for column in required_columns):
        raise ValueError("Evaluation fields must contain nonempty text.")

    test_cases = []
    for number, (_, row) in enumerate(df.iterrows(), start=1):
        answer, retrieved_chunks = rag(
            query=row["question"],
            k=k,
            model_name=MODEL,
            temperature=temperature,
            top_p=top_p,
            max_tokens=max_tokens,
            vectorstore=vectorstore,
        )
        relevant_context = [
            context.strip() for context in row["relevant_context"].split("<>")
            if context.strip()
        ]
        test_cases.append(LLMTestCase(
            input=row["question"],
            actual_output=answer,
            expected_output=row["expected_answer"],
            retrieval_context=retrieved_chunks,
            context=relevant_context,
        ))
        print(f"Prepared case {number}/{len(df)}", flush=True)
    return test_cases
````

For each row in the evaluation data, the function:

1. Runs the full RAG pipeline (retrieve + generate) on the question.
2. Splits the ground-truth relevant context on the `<>` delimiter, since some questions need multiple context passages.
3. Builds an `LLMTestCase` containing the input question, the generated answer, the expected answer, the retrieved context, and the ground-truth context.

This structure gives DeepEval everything it needs to score each metric.

## Running the Evaluation

The below step might take a few minutes because it runs the RAG pipeline on every question in the evaluation dataset. Each call involves embedding the query, searching the vector store, and generating an answer with the LLM.

````python
# Evaluate improved retrieval with the same generator and judge settings.
from pathlib import Path
from deepeval.evaluate import AsyncConfig, CacheConfig, DisplayConfig, ErrorConfig

K = 4
TEMPERATURE = 0.0
TOP_P = 1.0
MAX_TOKENS = 512
EVALUATION_OUTPUT_DIR = Path(".cache/notebook/evaluation")
EVALUATION_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

test_cases = test_case_generator(
    df, vectorstore, k=K, temperature=TEMPERATURE,
    top_p=TOP_P, max_tokens=MAX_TOKENS,
)
# Retain generated answers before starting judge calls.
(EVALUATION_OUTPUT_DIR / "test_cases.json").write_text(
    json.dumps([case.model_dump(mode="json") for case in test_cases], indent=2),
    encoding="utf-8",
)
results = evaluate(
    test_cases=test_cases,
    metrics=metrics,
    identifier="AuroraStays improved retrieval",
    hyperparameters={
        "model": MODEL, "judge": EVAL_MODEL, "k": K,
        "temperature": TEMPERATURE, "top_p": TOP_P, "max_tokens": MAX_TOKENS,
    },
    async_config=AsyncConfig(run_async=False),
    display_config=DisplayConfig(
        show_indicator=False, print_results=False, inspect_after_run=False,
        results_folder=str(EVALUATION_OUTPUT_DIR / "deepeval"),
    ),
    cache_config=CacheConfig(
        use_cache=os.environ.get("AURORASTAYS_EVALUATION_RESUME") == "1",
        write_cache=True,
    ),
    error_config=ErrorConfig(ignore_errors=False, skip_on_missing_params=False),
)
(EVALUATION_OUTPUT_DIR / "results.json").write_text(
    results.model_dump_json(indent=2), encoding="utf-8",
)
print(f"Scored {len(results.test_results)} cases across {len(metrics)} metrics.")
````

**Note:** The evaluation results above are primarily obtained using an LLM-as-a-Judge approach. Since LLM-based evaluation can be non-deterministic, the metric scores and observations may vary slightly across different evaluation runs.

Therefore, the results from a single run should not be treated as definitive. Running the evaluation multiple times and looking at the overall trends can provide a more reliable estimate of the system's performance.

**Note**: **If the results are not satisfactory, the previous code cell can be executed with a different set of parameter values, such as `k`, `top_p`, etc. Once the results are satisfactory, note down the parameter values.**

The function below converts raw evaluation results into a structured DataFrame with composite scores and routing decisions.

````python
def create_evaluation_df(test_cases, results):
    import math

    rows = []
    required_metrics = {
        "faithfulness", "contextual recall", "contextual precision", "answer relevancy"
    }
    for test_case, result in zip(test_cases, results.test_results, strict=True):
        if result.input != test_case.input:
            raise ValueError("Evaluation results do not match the test-case order.")
        metric_data = result.metrics_data or []
        if any(metric.error for metric in metric_data):
            raise ValueError("A judge metric failed; inspect the saved raw results.")
        scores = {
            metric.name.replace(" [GEval]", "").strip().lower(): metric.score
            for metric in metric_data
        }
        if set(scores) != required_metrics or len(metric_data) != len(required_metrics):
            raise ValueError("Each case must have exactly the four required metrics.")
        if not all(isinstance(score, (int, float)) and math.isfinite(score) and 0 <= score <= 1
                   for score in scores.values()):
            raise ValueError("Metric scores must be finite numbers between 0 and 1.")
        composite = (
            0.35 * scores["faithfulness"]
            + 0.25 * scores["contextual recall"]
            + 0.20 * scores["contextual precision"]
            + 0.20 * scores["answer relevancy"]
        )
        rows.append({
            "Question": test_case.input,
            "Faithfulness Score": scores["faithfulness"],
            "Contextual Recall Score": scores["contextual recall"],
            "Contextual Precision Score": scores["contextual precision"],
            "Answer Relevancy Score": scores["answer relevancy"],
            "Composite Score": composite,
            "Type": "AUTO ANSWER" if composite >= 0.75 else "MANUAL LOOKUP",
            "Generated Answer": test_case.actual_output,
            "Expected Answer": test_case.expected_output,
        })
    return pd.DataFrame(rows)
````

The composite score uses the business weights: Faithfulness 35%, Contextual Recall 25%, Contextual Precision 20%, and Answer Relevancy 20%.

The benchmark routing rule matches the success criteria at the start of the notebook:
- **AUTO ANSWER:** composite score >= 0.75.
- **MANUAL LOOKUP:** composite score < 0.75.

Individual metric pass counts and the >= 50% auto-answer target are reported separately. These are offline evaluation labels: contextual recall depends on an expected answer, so the scores do not provide a calibrated confidence measure for unseen live queries.


````python
# Display scores, metric pass counts, and the benchmark auto-answer rate.
evaluation_df = create_evaluation_df(test_cases, results)
display(evaluation_df)
metric_targets = {
    "Faithfulness": 0.85,
    "Contextual Recall": 0.80,
    "Contextual Precision": 0.75,
    "Answer Relevancy": 0.75,
}
metric_summary = pd.DataFrame([
    {
        "Metric": name,
        "Mean Score": float(evaluation_df[f"{name} Score"].mean()),
        "Target": target,
        "Passing Cases": int((evaluation_df[f"{name} Score"] >= target).sum()),
        "Total Cases": len(evaluation_df),
    }
    for name, target in metric_targets.items()
])
display(metric_summary)
auto_answer_count = int((evaluation_df["Type"] == "AUTO ANSWER").sum())
auto_answer_rate = auto_answer_count / len(evaluation_df)
print(f"AUTO ANSWER: {auto_answer_count}/{len(evaluation_df)} ({auto_answer_rate:.0%}); target >= 50%")
print(f"MANUAL LOOKUP: {len(evaluation_df) - auto_answer_count}/{len(evaluation_df)}")
print("These are offline benchmark routes, not calibrated confidence scores for new guest queries.")

evaluation_df.to_csv(EVALUATION_OUTPUT_DIR / "scores.csv", index=False)
metric_summary.to_csv(EVALUATION_OUTPUT_DIR / "metric_summary.csv", index=False)
(EVALUATION_OUTPUT_DIR / "summary.json").write_text(json.dumps({
    "dataset": "Data/evaluation.csv",
    "cases": len(evaluation_df),
    "model": MODEL,
    "judge": EVAL_MODEL,
    "parameters": {"k": K, "temperature": TEMPERATURE, "top_p": TOP_P, "max_tokens": MAX_TOKENS},
    "metric_summary": metric_summary.to_dict(orient="records"),
    "mean_composite": float(evaluation_df["Composite Score"].mean()),
    "auto_answer_count": auto_answer_count,
    "auto_answer_rate": auto_answer_rate,
    "auto_answer_rate_target": 0.50,
    "auto_answer_rate_target_met": auto_answer_rate >= 0.50,
    "routing_rule": "AUTO ANSWER at composite >= 0.75; MANUAL LOOKUP below 0.75",
    "all_metrics_passing_cases": int(evaluation_df[[f"{name} Score" for name in metric_targets]].ge(list(metric_targets.values())).all(axis=1).sum()),
    "test_set_evaluated": False,
}, indent=2) + "\n", encoding="utf-8")
print(f"Evaluation artifacts saved to: {EVALUATION_OUTPUT_DIR}")
````

- The output table shows each evaluation question alongside its individual metric scores, the weighted composite score, and the routing decision.
- This gives a clear picture of where the system is performing well and where it needs improvement.
- Questions routed to "AUTO ANSWER" are the ones the system can handle confidently without human intervention.

Its `rag` function uses the same defaults, timeout, and retry setting, with credentials read from environment variables for deployment. The packaged `policy_retrieval.py` keeps structured parent contexts and hybrid ranking identical to the notebook.


````python
%%writefile backend_files/rag_utils.py
"""Evaluated notebook RAG functions; server credentials come from environment variables."""

import os
from langchain_openai import ChatOpenAI

# Combine keyword matching and FAISS, returning up to k distinct policy contexts.
from policy_retrieval import retrieve_policy_documents

def retrieve(query: str, k: int = 4, vectorstore=None) -> list[str]:
    results = retrieve_policy_documents(vectorstore, query, k=k)
    return [
        f"Content:\n{doc.page_content}\n\nMetadata:\n" + str({
            "source": doc.metadata["source"],
            "sources": doc.metadata.get("sources", [doc.metadata["source"]]),
            "section": doc.metadata.get("heading", ""),
        })
        for doc in results
    ]

# Define the generation function that prompts the LLM with the retrieved context.
def generate(query, retrieved_chunks, model):
    rules = """
    - Use only the retrieved policy information to answer the question.
    - Match the requested property exactly. Preserve fee amounts, currencies,
      per-night/per-stay units, caps, conditions, and exceptions.
    - If the information needed to answer is present, give a direct, concise answer.
      Other properties' rows do not override the requested property's row.
    - Cite the source filename from the metadata of the chunk containing each
      fact used in the answer. Do not cite a different chunk about another property.
    - If the retrieved information is insufficient or conflicting, explain what
      is missing or conflicting and ask for verification with the current property
      profile or property team. Do not invent policy details or use outside knowledge.
    """
    prompt = f"""
    You are an AI assistant answering questions about hotel policies.

    User Query:
    {query}

    Retrieved Policy Information:
    {retrieved_chunks}

    Rules:
    {rules}

    Answer:
    """
    return model.invoke(prompt)

# Combine retrieval and generation, returning answer text and source excerpts.
def rag(
    query: str,
    k: int = 4,
    model_name: str = "gpt-4o-mini",
    temperature: float = 0.0,
    top_p: float = 1.0,
    max_tokens: int = 512,
    vectorstore=None,
):
    if vectorstore is None:
        raise ValueError("Provide the initialized FAISS vectorstore.")

    model = ChatOpenAI(
        model=model_name,
        temperature=temperature,
        top_p=top_p,
        max_tokens=max_tokens,
        api_key=os.environ["OPENAI_API_KEY"],
        base_url=os.environ["OPENAI_API_BASE"].strip(),
        timeout=60,
        max_retries=0,
    )
    retrieved_chunks = retrieve(query=query, k=k, vectorstore=vectorstore)

    answer = generate(query=query, retrieved_chunks=retrieved_chunks, model=model)
    return answer.content, retrieved_chunks
````

The functions are written to `backend_files/rag_utils.py`. Set `OPENAI_API_KEY` and `OPENAI_API_BASE` in the server environment. The local launcher can load ignored `config.json` into its own process environment; credentials are never copied into the backend package.


The Flask application loads the normalized embedding model and verified index once per worker. Its factory also accepts a vector store for offline endpoint tests. The three API paths and response formats are preserved, with `k=4` and the evaluated generation settings as defaults.


````python
%%writefile backend_files/app.py
"""Flask endpoints using the notebook's evaluated hotel-policy pipeline."""

import hashlib
import json
import math
from pathlib import Path

from flask import Flask, jsonify, request
from rag_utils import rag, retrieve

BACKEND_DIR = Path(__file__).resolve().parent


def load_vectorstore():
    from langchain_community.vectorstores import FAISS
    from langchain_huggingface import HuggingFaceEmbeddings

    manifest = json.loads((BACKEND_DIR / "deployment_manifest.json").read_text())
    # Verify our own exported index before loading its pickle-based docstore.
    for relative, expected in manifest["artifact_hashes"].items():
        if hashlib.sha256((BACKEND_DIR / relative).read_bytes()).hexdigest() != expected:
            raise RuntimeError(f"Backend artifact does not match its manifest: {relative}")
    embeddings = HuggingFaceEmbeddings(
        model_name=manifest["embedding_model"],
        model_kwargs={"device": "cpu", "revision": manifest["embedding_revision"]},
        encode_kwargs={"normalize_embeddings": True},
    )
    vectorstore = FAISS.load_local(
        str(BACKEND_DIR / "faiss_index"), embeddings,
        allow_dangerous_deserialization=True,
    )
    if (vectorstore.index.ntotal, vectorstore.index.d) != (manifest["vectors"], manifest["dimensions"]):
        raise RuntimeError("Backend index dimensions/count differ from the evaluated index.")
    return vectorstore


def request_parameters():
    data = request.get_json(silent=True)
    if not isinstance(data, dict):
        raise ValueError("Send a JSON object containing query.")
    query = data.get("query")
    if not isinstance(query, str) or not query.strip():
        raise ValueError("query must be nonempty text.")
    if len(query) > 8000:
        raise ValueError("query must contain at most 8000 characters.")
    k = data.get("k", 4)
    if type(k) is not int or not 1 <= k <= 16:
        raise ValueError("k must be an integer from 1 to 16.")
    model_name = data.get("model_name", "gpt-4o-mini")
    if not isinstance(model_name, str) or not model_name.strip() or len(model_name) > 128:
        raise ValueError("model_name must be nonempty text of at most 128 characters.")
    parameters = {"query": query.strip(), "k": k, "model_name": model_name.strip()}
    for name, default, lower, upper in (("temperature", 0.0, 0.0, 2.0), ("top_p", 1.0, 0.0, 1.0)):
        value = data.get(name, default)
        if type(value) not in (int, float) or not math.isfinite(value) or not lower <= value <= upper:
            raise ValueError(f"{name} must be a finite number from {lower} to {upper}.")
        if name == "top_p" and value == 0:
            raise ValueError("top_p must be greater than zero.")
        parameters[name] = value
    max_tokens = data.get("max_tokens", 512)
    if type(max_tokens) is not int or not 1 <= max_tokens <= 4096:
        raise ValueError("max_tokens must be an integer from 1 to 4096.")
    parameters["max_tokens"] = max_tokens
    return parameters


def create_app(vectorstore=None):
    """Load the store once per worker, or accept a store for offline tests."""
    rag_api = Flask("Aurora Stay API")
    rag_api.config["MAX_CONTENT_LENGTH"] = 64 * 1024
    store = load_vectorstore() if vectorstore is None else vectorstore
    rag_api.config["VECTORSTORE"] = store

    @rag_api.get("/")
    def home():
        return "Welcome to the Aurora Stay RAG API!"

    @rag_api.post("/v1/relevant_chunks")
    def relevant_chunks():
        try:
            parameters = request_parameters()
        except ValueError as error:
            return jsonify({"error": str(error)}), 400
        chunks = retrieve(parameters["query"], k=parameters["k"], vectorstore=store)
        return jsonify({"query": parameters["query"], "k": parameters["k"], "relevant_chunks": chunks})

    def answer_response(include_chunks):
        try:
            parameters = request_parameters()
        except ValueError as error:
            return jsonify({"error": str(error)}), 400
        try:
            answer, chunks = rag(**parameters, vectorstore=store)
        except Exception as error:
            # Provider exception messages can contain request/credential details.
            rag_api.logger.warning("Answer generation failed (%s).", type(error).__name__)
            return jsonify({"error": "Answer generation failed. Please retry or contact the property team."}), 502
        payload = {"answer": answer}
        if include_chunks:
            payload.update(parameters, relevant_chunks=chunks)
        return jsonify(payload)

    @rag_api.post("/v1/answer_with_relevant_chunks")
    def answer_with_relevant_chunks():
        return answer_response(include_chunks=True)

    @rag_api.post("/v1/answer")
    def answer():
        return answer_response(include_chunks=False)

    return rag_api


if __name__ == "__main__":
    create_app().run(host="127.0.0.1", port=7860, debug=False)
````

The Flask API provides `/v1/relevant_chunks`, `/v1/answer_with_relevant_chunks`, and `/v1/answer`. Requests require a nonempty JSON `query`; parameter validation returns HTTP 400 and upstream generation failures return a sanitized HTTP 502 response. The application resolves its index relative to `app.py`, verifies its manifest hashes, and loads only this locally exported pickle-based docstore. This API generates answers; benchmark GEval scores and AUTO ANSWER routing are not computed for new guest queries.

The policy desk lets hotel staff ask booking, room, amenity, fee, and guest-policy questions. It connects to the backend through a server-side `BACKEND_URL` setting, using the evaluated defaults. API credentials remain in the backend.


````python
%%writefile frontend_files/app.py
"""Staff-facing hotel policy questions, using the evaluated backend defaults."""

import ast
import os

import requests
import streamlit as st

BACKEND_URL = os.environ.get("BACKEND_URL", "http://127.0.0.1:7860").strip().rstrip("/")
REQUEST_TIMEOUT = (5, 90)


def request_answer(query, include_sources, k, temperature, max_tokens):
    endpoint = "/v1/answer_with_relevant_chunks" if include_sources else "/v1/answer"
    response = requests.post(
        BACKEND_URL + endpoint,
        json={"query": query, "k": k, "model_name": "gpt-4o-mini",
              "temperature": temperature, "top_p": 1.0, "max_tokens": max_tokens},
        timeout=REQUEST_TIMEOUT,
    )
    response.raise_for_status()
    result = response.json()
    if not isinstance(result, dict) or not isinstance(result.get("answer"), str) or not result["answer"].strip():
        raise ValueError("Invalid answer response.")
    chunks = result.get("relevant_chunks", [])
    if include_sources and ("relevant_chunks" not in result or not isinstance(chunks, list)
                            or not all(isinstance(chunk, str) for chunk in chunks)):
        raise ValueError("Invalid policy excerpts.")
    return {"query": query, "answer": result["answer"].strip(),
            "chunks": chunks if include_sources else [], "include_sources": include_sources}


def policy_excerpt(chunk):
    content, _, raw_metadata = chunk.partition("\n\nMetadata:\n")
    content = content.removeprefix("Content:\n").strip()
    try:
        metadata = ast.literal_eval(raw_metadata) if raw_metadata else {}
    except (SyntaxError, ValueError):
        metadata = {}
    if not isinstance(metadata, dict):
        metadata = {}
    heading = metadata.get("section", "")
    if content.startswith("Section: "):
        first, _, content = content.partition("\n")
        heading = heading or first.removeprefix("Section: ")
    sources = metadata.get("sources") or [metadata.get("source", "")]
    if not isinstance(sources, list):
        sources = []
    filenames = list(dict.fromkeys(source.replace("\\", "/").rsplit("/", 1)[-1]
                                   for source in sources if isinstance(source, str) and source))
    return content.strip(), heading, filenames


st.set_page_config(page_title="Aurora Stays | Policy desk", page_icon="🏨", layout="centered")
st.title("🏨 Aurora Stays")
st.write("Answers to your hotel policy questions.")

with st.sidebar:
    st.header("Policy desk")
    st.caption("Ask about bookings, rooms, amenities, fees, or guest policies.")
    with st.expander("Advanced answer settings", expanded=False):
        st.caption("The defaults match the evaluated configuration.")
        k = st.slider("Policy excerpts to retrieve", 1, 10, 4, key="retrieval_k")
        temperature = st.slider("Temperature", 0.0, 1.0, 0.0, 0.1, key="temperature")
        max_tokens = st.slider("Answer token limit", 64, 2048, 512, 64, key="max_tokens")

with st.form("policy_question", clear_on_submit=False):
    query = st.text_area(
        "Your question", placeholder="What are the pet fees at ASH-CHI Loop?",
        height=110, max_chars=8000, key="question",
    )
    response_mode = st.radio(
        "Response mode", ["Answer only", "Answer with sources"],
        horizontal=True, key="response_mode",
    )
    submitted = st.form_submit_button("Ask", type="primary", width="stretch")

st.session_state.setdefault("policy_result", None)
if submitted:
    st.session_state["policy_result"] = None
    if not query.strip():
        st.warning("Please enter a hotel policy question.")
    else:
        try:
            with st.spinner("Checking the policy documents…"):
                st.session_state["policy_result"] = request_answer(
                    query.strip(), response_mode == "Answer with sources", k, temperature, max_tokens,
                )
        except requests.Timeout:
            st.error("The answer is taking longer than expected. Please try again.")
        except requests.HTTPError:
            st.error("The policy service could not answer this request. Please try again shortly.")
        except requests.RequestException:
            st.error("The policy service is unavailable. Please try again shortly.")
        except (ValueError, TypeError):
            st.error("The policy service returned an unexpected response. Please try again.")

result = st.session_state["policy_result"]
if result:
    st.divider()
    st.subheader("Answer")
    st.caption(result["query"])
    st.markdown(result["answer"])
    if result["include_sources"]:
        st.subheader("Supporting policy excerpts")
        if not result["chunks"]:
            st.info("No supporting excerpts were returned for this question.")
        for number, chunk in enumerate(result["chunks"], 1):
            content, heading, sources = policy_excerpt(chunk)
            with st.expander(heading or f"Policy excerpt {number}"):
                st.markdown(content.replace("\n", "  \n"))
                if sources:
                    st.caption("Sources: " + ", ".join(sources))
````

The frontend keeps the two response modes: Answer only and Answer with sources. Advanced settings retain adjustable retrieval count, temperature, and output-token limit; defaults are K=4, temperature=0 and 512 tokens, with gpt-4o-mini and top_p=1 in requests. A form prevents requests while typing or changing settings, and session state preserves the last answer on reruns. Supporting excerpts show policy text and source filenames. Connection, timeout, HTTP, and malformed-response errors are handled without exposing internal request details.
